```python
from typing import List


def binary_search(data: List[int], target: int) -> int:
    low = 0
    high = len(data) - 1

    while low <= high:
        mid = (low + high) // 2

        if data[mid] == target:
            return mid
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1


if __name__ == "__main__":
    data = [10, 20, 30, 40, 50, 60, 70]
    target = 40

    result = binary_search(data, target)

    print("Data:", data)
    print("Target:", target)
    print("Result index:", result)

    if result != -1:
        print("Inference: Target element is present.")
    else:
        print("Inference: Target element is not present.")
```
